```python
train_data.Age=train_data.Age.fillna(train_data.Age.mean())
```

```python
train_data.drop('Cabin', axis=1, inplace=True, errors='ignore')
```

```python
# Fill Embarked with mode
train_data['Embarked'] = train_data['Embarked'].fillna(train_data['Embarked'].mode()[0])
```

```python
train_data.isnull().sum()
```

```python
test_data.isnull().sum()
```

```python
# Fill Age with median
test_data['Age'] = test_data['Age'].fillna(test_data['Age'].median())

# Drop Cabin column (optional)
test_data.drop('Cabin', axis=1, inplace=True, errors='ignore')
```

```python
test_data['Fare'] = test_data['Fare'].fillna(test_data['Fare'].median())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Univariate Analysis: Distribution of 'Age'
plt.figure(figsize=(8, 5))
sns.histplot(train_data['Age'], kde=True, bins=30, color='blue')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
# Univariate Analysis: Distribution of 'Fare'
plt.figure(figsize=(8, 5))
sns.histplot(train_data['Fare'], kde=True, bins=30, color='green')
plt.title('Fare Distribution of Passengers')
plt.xlabel('Fare')
plt.ylabel('Count')
plt.show()
```

```python
# Univariate Analysis: Survival Count
plt.figure(figsize=(8, 5))
sns.countplot(x='Survived', data=train_data)
plt.title('Survival Count (0: No, 1: Yes)')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

<h4>Univariate Analysis:

Focuses on single variables like Age, Fare, and Survival to understand their distribution and basic statistics.

The countplot for Survival helps us see the class distribution (survived vs. not survived).

</h4>

```python
# Bivariate Analysis: Survival by Gender
plt.figure(figsize=(8, 5))
sns.countplot(x='Sex', hue='Survived', data=train_data)
plt.title('Survival by Gender')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()

```

```python
# Bivariate Analysis: Survival by Passenger Class
plt.figure(figsize=(8, 5))
sns.countplot(x='Pclass', hue='Survived', data=train_data)
plt.title('Survival by Passenger Class')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()

```

```python
# Bivariate Analysis: Age vs Survival (Boxplot)
plt.figure(figsize=(8, 5))
sns.boxplot(x='Survived', y='Age', data=train_data)
plt.title('Age vs Survival')
plt.xlabel('Survived')
plt.ylabel('Age')
plt.show()

```

```python
# Bivariate Analysis: Fare vs Survival (Boxplot)
plt.figure(figsize=(8, 5))
sns.boxplot(x='Survived', y='Fare', data=train_data)
plt.title('Fare vs Survival')
plt.xlabel('Survived')
plt.ylabel('Fare')
plt.show()
```

<h4>Bivariate Analysis:

Looks at relationships between two variables at a time (e.g., Survival vs Gender, Survival vs Class).

Boxplots and countplots allow us to compare distributions and survival rates across categorical variables.</h4>

```python
# Multivariate Analysis: Pairplot (Age, Fare, Pclass, Survived)
sns.pairplot(train_data[['Age', 'Fare', 'Pclass', 'Survived']], hue='Survived')
plt.title('Pairplot: Age, Fare, Pclass, and Survival')
plt.show()

```

```python
# Multivariate Analysis: Heatmap (Correlation between numerical features)
plt.figure(figsize=(10, 6))
numeric_cols = train_data.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python

# Multivariate Analysis: 3D plot (Age, Fare, and Survival)
from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(train_data['Age'], train_data['Fare'], train_data['Survived'], c=train_data['Survived'], cmap='viridis')
ax.set_xlabel('Age')
ax.set_ylabel('Fare')
ax.set_zlabel('Survival')
ax.set_title('3D Plot: Age, Fare, and Survival')
plt.show()
```

<h4>Multivariate Analysis:

Analyzes interactions between multiple variables at once (e.g., Age, Fare, Pclass, and Survival).

Pairplots show the relationships between several variables, while heatmaps highlight correlations.</h4>
